```python
import networkx as nx
import matplotlib.pyplot as plt
```

### Create a directed graph

```python
# Create a directed graph
G = nx.DiGraph()
# Add nodes
G.add_node(1)
G.add_node(2)
G.add_node(3)
G.add_node(4)
G.add_node(5)
G.add_node(6)
# Add edges
G.add_edge(1, 2)
G.add_edge(2, 3)
G.add_edge(3, 4)
G.add_edge(4, 1)
G.add_edge(1, 4)
G.add_edge(4, 5)
G.add_edge(5, 6)
```

```python
G
```

```python
# plot the graph
nx.draw(G, with_labels=True, node_size=2000, node_color="skyblue", font_size=10, font_color="black", font_weight="bold")
# show the plot
plt.show()
```

```python
# calculate the path between two nodes
for shortest_path in nx.all_shortest_paths(G, 1, 5):
    print('Shortest path between 1 and 5: ', shortest_path)
    # calculate the shortest path length
    print('Shortest path length between 1 and 5: ', len(shortest_path) - 1)
# or using built-in function
print('Shortest path length between 1 and 5: ',nx.shortest_path_length(G, 1, 5))
```

### Create a indirected graph

```python
# build a in-directed graph
G = nx.Graph(directed=False)
G.add_node(1)
G.add_node(2)
G.add_node(3)
G.add_node(4)
G.add_node(5)
G.add_node(6)
# Add edges
G.add_edge(1, 2)
G.add_edge(2, 3)
G.add_edge(3, 4)
G.add_edge(4, 1)
G.add_edge(4, 5)
G.add_edge(5, 6)
```

```python
G
```

```python
# plot the graph
nx.draw(G, with_labels=True, node_size=2000, node_color="skyblue", font_size=10, font_color="black", font_weight="bold")
# show the plot
plt.show()
```

```python
# calculate the path between two nodes
for possible_path in nx.all_simple_paths(G, 2, 5):
    print('Possible path between 2 and 5: ', possible_path)
    # calculate path length
    print('Path length between 2 and 5: ', len(possible_path)-1)
```

### Weighted Indirected Graph

In above case, there are exist two shortest path between node 2 and 5 which are 2->1->4->5 and 2->3->4->5. 

This based on the assumption that all the path with save weight. In this example, we can adjust the weight of the edge between node 1 and 2 or 2 and 3, then see the result.

```python
# Build a weighted graph
G = nx.Graph()
G.add_node(1)
G.add_node(2)
G.add_node(3)
G.add_node(4)
G.add_node(5)
G.add_node(6)
# Add weighted edges
G.add_edge(1, 2, weight=1)
# adjust the weight of the edge between 2 and 3
G.add_edge(2, 3, weight=3)
G.add_edge(3, 4, weight=1)
G.add_edge(4, 1, weight=1)
G.add_edge(4, 5, weight=1)
G.add_edge(5, 6, weight=1)
# add weight to the edge between 2 and 5
```

```python
# plot the graph

# positions for all nodes - seed for reproducibility
pos = nx.spring_layout(G, seed=7)

# nodes
nx.draw_networkx_nodes(G, pos, node_size=700, node_color = "skyblue")

# edges
nx.draw_networkx_edges(G, pos)



# node labels
nx.draw_networkx_labels(G, pos, font_size=10, font_family="sans-serif")
# edge weight labels
edge_labels = nx.get_edge_attributes(G, "weight")
nx.draw_networkx_edge_labels(G, pos, edge_labels)

ax = plt.gca()
ax.margins(0.08)
plt.axis("off")
plt.tight_layout()

# show the plot
plt.show()
```

```python
# calculate the shortest path between two nodes
for shortest_path in nx.all_shortest_paths(G, 2, 5, weight='weight'):
    print('Shortest path between 2 and 5: ', shortest_path)
    # calculate the shortest path length
    print('Shortest path length between 2 and 5: ', len(shortest_path) - 1)
```

#### Insertion in Binary Search Trees (BST)

Let's randomly generate a bst with height 2, not perfect.

```python
binary_search_tree_1 = bst(height=2, is_perfect=False)
print(binary_search_tree_1)
```

**Question:** If we wish to insert key = 4, then what is the approach?

A standard BST insertion follows these steps:

1. Start at the root.
2. Compare the key with the current node value.
3. If key is smaller, move to the left child; if larger, move to the right child.
4. Repeat until you find an empty position (`None`).
5. Insert the new key at that position.

For key `4`, you keep comparing and moving left/right until a `None` child is found, then place `4` there.

Discuss this decision path in the tutorial session and present it on the board.

```python
def bst_insert(node, key):
    if node is None:
        return Node(key)

    if key < node.value:
        node.left = bst_insert(node.left, key)
    elif key > node.value:
        node.right = bst_insert(node.right, key)
    else:
        # If key already exists, do nothing
        return node

    return node

print("Tree before inserting 4:")
print(binary_search_tree_1)

binary_search_tree_1 = bst_insert(binary_search_tree_1, 4)

print("\nTree after inserting 4:")
print(binary_search_tree_1)
```

```python
binary_search_tree_2 = bst_insert(binary_search_tree_1, 9)
print("\nTree after inserting 3:")
print(binary_search_tree_2)
```

### Usage Case 1: Route Planning with Weighted Graphs

A weighted graph can represent roads or transport links, where each edge weight is travel time.
The shortest path then gives the most efficient route under that cost definition.

```python
# Example: travel-time network between locations
route_graph = nx.Graph()

route_graph.add_weighted_edges_from([
    ("Campus", "Station", 12),
    ("Campus", "Library", 6),
    ("Library", "Station", 4),
    ("Station", "Mall", 8),
    ("Library", "Mall", 15),
    ("Mall", "Hospital", 7),
])

best_route = nx.shortest_path(route_graph, source="Campus", target="Hospital", weight="weight")
best_time = nx.shortest_path_length(route_graph, source="Campus", target="Hospital", weight="weight")

print("Best route:", " -> ".join(best_route))
print("Estimated travel time:", best_time, "minutes")

# Which location appears most connected?
centrality = nx.degree_centrality(route_graph)
most_connected = max(centrality, key=centrality.get)
print("Most connected location:", most_connected, "(centrality=", round(centrality[most_connected], 3), ")")
```

### Usage Case 2: Traversal and Search in a Binary Tree

Tree traversal is useful for checking structure, exporting values in a specific order,
and implementing search-related logic.

```python
# Build a binary tree from list data and inspect different traversal orders
example_values = [10, 5, 15, 2, 7, 12, 20]
example_tree = build(example_values)
print(example_tree)

def inorder(node):
    if node is None:
        return []
    return inorder(node.left) + [node.value] + inorder(node.right)

def preorder(node):
    if node is None:
        return []
    return [node.value] + preorder(node.left) + preorder(node.right)

def contains_value(node, target):
    if node is None:
        return False
    if node.value == target:
        return True
    return contains_value(node.left, target) or contains_value(node.right, target)

print("In-order traversal:", inorder(example_tree))
print("Pre-order traversal:", preorder(example_tree))
print("Contains 12?", contains_value(example_tree, 12))
print("Contains 99?", contains_value(example_tree, 99))
```

### Usage Case 3: Converting Nested Data to a General Tree

Nested dictionaries (similar to JSON documents) are common in APIs.
Mapping them into a tree helps with structure discovery and traversal.

```python
# Example nested data (API-like JSON structure)
from anytree import Node as AnyNode

nested_data = {
    "University": {
        "Faculty of IT": {
            "Units": ["FIT5196", "FIT9132"],
            "Programs": ["Data Science", "Business Information Systems"],
        },
        "Library": {
            "Services": ["Borrowing", "Study Rooms"]
        },
    }
}

def dict_to_anytree(name, obj, parent=None):
    current = AnyNode(name, parent=parent)
    if isinstance(obj, dict):
        for key, value in obj.items():
            dict_to_anytree(key, value, parent=current)
    elif isinstance(obj, list):
        for item in obj:
            dict_to_anytree(str(item), {}, parent=current)
    else:
        AnyNode(str(obj), parent=current)
    return current

hierarchy_root = dict_to_anytree("root", nested_data)
for pre, _, node in RenderTree(hierarchy_root):
    print(f"{pre}{node.name}")
```
